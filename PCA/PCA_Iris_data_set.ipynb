```python
from sklearn.datasets import load_iris

data = load_iris()
X = data.data
y = data.target

#We can already work with the data
```

```python
import numpy as np

# Center the matrix
X_centered = X-X.mean(axis=0)

# Covariance matrix
X_cov = (1/(len(X)-1))*np.matmul(X_centered , X_centered.T)

# Find eigenvalues and vector
eigenvalues, eigenvector = np.linalg.eigh(X_cov)

# Sorting eigenvalues and eigenvectors
idx = eigenvalues.argsort()[::-1]   
eigenValues = eigenvalues[idx]
eigenVectors = eigenvector[:,idx]

# Explained variance
explained_variance_ratio = eigenValues / np.sum(eigenValues)

# Projection
np.multiply(X,eigenValues)
```
